```python
# Calculate daily returns
prices = stock.history(period='126mo')
prices['Returns'] = prices['Close'].pct_change()

# Group by year and sum returns
annual_returns = prices['Returns'].groupby(prices.index.year).sum()

# Convert to DataFrame for easier viewing
annual_returns_df = annual_returns.reset_index()
annual_returns_df.columns = ['Year', 'Annual_Return']

# If you want it as a dictionary
annual_returns_dict = annual_returns.to_dict()

# If you need specific years as separate variables
for year, ret in annual_returns.items():
    print(f"Year {year}: {ret:.4%}")
```
